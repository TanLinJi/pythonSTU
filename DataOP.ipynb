#### N维数组是机器学习和神经网络的主要数据结构

创建数组
1. 形状
2. 每个元素的数据类型
3. 每个元素的值

```python
import torch
```

张量表示一个数值组成的数组，这个数组可能有多个维度

```python
x = torch.arange(12)
x
```

```python
x.shape  # 通过shape来访问张量的形状
```

```python
x.numel() # 通过numel来访问张量元素的个数
```

通过reshape来改变一个张量的形状而不改变元素数量和元素值

```python
x = x.reshape(3,4)
x
```

使用全0，全1，其他常量或者从特定分布中随机采样的数字

```python
torch.zeros((2,3,4))  # 创建了一个全0的三维数组
```

```python
torch.ones((2,3,4))  # 创建了一个全1的三维数组
```

通过提供包含数值的Python列表（或嵌套列表）来为所需张量中的每个元素赋予确定值

```python
torch.tensor([[2,1,4,3],[1,2,3,4],[4,3,2,1]]) # 列表嵌套列表 -->二维数组
```

```python
torch.tensor([[[2,1,4,3],[1,2,3,4],[4,3,2,1]]]) 
```

```python
torch.tensor([[[2,1,4,3],[1,2,3,4],[4,3,2,1]]]).shape
```

常见的标准算术运算符（+, -, *, /, **） 都可以被升级为按元素运算

```python
x = torch.tensor([1.0, 2, 4, 8])
y = torch.tensor([2, 2, 2, 2])
x,y
```

```python
x + y
```

```python
x - y
```

```python
x * y
```

```python
x / y
```

```python
x ** y
```

```python
torch.exp(x)  # 可以按元素执行更多运算，eg.按指数
```

把多个张量连结在一起

```python
X = torch.arange(12,dtype=torch.float32).reshape((3,4))
Y = torch.tensor([[2.0,1,4,3],[1,2,3,4],[4,3,2,1]])
X,Y
```

```python
torch.cat((X,Y),dim=0)  # dim = 0, 表示在列上做拼接，竖着拼
```

```python
torch.cat((X,Y),dim=1) # dim = 0, 表示在行上做拼接，横着拼
```

通过逻辑运算符来构建二维的张量

```python
X == Y
```

对张量中的所有元素求和会产生一个只有一个元素的张量

```python
X.sum()
```

对于形状不同的张量，可以通过调用 *广播机制*  来执行按元素的操作

```python
a = torch.arange(3).reshape((3,1))
b = torch.arange(2).reshape((1,2))
a,b
```

```python
a + b  # 背后是把 a 的那一列复制一列，复制成一个3*2的矩阵
       # 把 b 的那一行复制两行，也复制成一个 3*2 的矩阵
```

##### 访问元素

```python
X[-1]   # 使用 -1 访问最后一个元素
```

```python
X[1:3]  # 使用1：3访问第二个和第三个元素
```

指定索引来将元素写入矩阵

```python
X
```

```python
X[1, 2] = 99
X
```

为多个元素指定相同的值

```python
X[0:2, :] = 12
X
```
