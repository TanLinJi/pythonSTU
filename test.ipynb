### eval函数的使用

- eval函数用于去掉字符串左右的引号
- eval函数经常和Input函数组合使用，用于获得输入的字符型数据

```python
s = '3.14 + 3'
print(s,type(s))

x = eval(s)  # 使用eval函数去掉s这个字符串中左右的引号，执行加法运算
print(x, type(x))
```

```python
age = eval(input('Please tell me your age: '))
print(age,type(age))
```

## 算术运算符

| 运算符 | 描述                                               | 示例      | 结果            |                          |
| ------ | -------------------------------------------------- | --------- | --------------- | ------------------------ |
| `+`    | 加法：对两个操作数进行求和                         | `5 + 3`   | `8`             |                          |
| `-`    | 减法：用第一个操作数减去第二个操作数               | `10 - 4`  | `6`             |                          |
| `*`    | 乘法：对两个操作数进行相乘                         | `6 * 7`   | `42`            |                          |
| `/`    | 除法：用第一个操作数除以第二个操作数，结果为浮点数 | `15 / 4`  | `3.75`          |  除数不允许为0             |
| `//`   | 地板除法（取整除）：返回除法的整数部分，向下取整   | `15 // 4` | `3`；`-15 // 4` | `-4`                     |
| `%`    | 取模（求余数）：返回除法的余数                     | `15 % 4`  | `3`；`15 % -4`  | `-1`                     |
| `**`   | 幂运算：计算第一个操作数的第二个操作数次幂         | `2** 3`   | `8`；`3 **0.5`  | `1.732...`（3 的平方根） |

算术运算符的优先级：幂运算>乘除>加减

## 赋值运算符
| 运算符 | 描述                                                         | 示例      | 展开形式     |
| ------ | ------------------------------------------------------------ | --------- | ------------ |
| `=`    | 基本赋值：将右侧值赋给左侧变量                               | `x = 5`   | `x = 5`      |
| `+=`   | 加法赋值：将右侧值加到左侧变量后再赋值给左侧                 | `x += 3`  | `x = x + 3`  |
| `-=`   | 减法赋值：将左侧变量减去右侧值后再赋值给左侧                 | `x -= 2`  | `x = x - 2`  |
| `*=`   | 乘法赋值：将左侧变量乘以右侧值后再赋值给左侧                 | `x *= 4`  | `x = x * 4`  |
| `/=`   | 除法赋值：将左侧变量除以右侧值后再赋值给左侧                 | `x /= 2`  | `x = x / 2`  |
| `//=`  | 整除法赋值：将左侧变量对右侧值做整除法后再赋值给左侧     | `x //= 3` | `x = x // 3` |
| `%=`   | 取模赋值：将左侧变量对右侧值取模后再赋值给左侧               | `x %= 5`  | `x = x % 5`  |
| `**=`  | 幂运算赋值：将左侧变量对右侧值做幂运算后再赋值给左侧         | `x**= 3`  | `x = x ** 3` |
| `&=`   | 按位与赋值：将左侧变量与右侧值做按位与运算后再赋值给左侧     | `x &= 3`  | `x = x & 3`  |
| `^=`   | 按位异或赋值：将左侧变量与右侧值做按位异或运算后再赋值给左侧 | `x ^= 2`  | `x = x ^ 2`  |
| `<<=`  | 左移赋值：将左侧变量左移右侧值指定的位数后再赋值给左侧       | `x <<= 2` | `x = x << 2` |
| `>>=`  | 右移赋值：将左侧变量右移右侧值指定的位数后再赋值给左侧       | `x >>= 1` | `x = x >> 1` |

```python
# python支持链式赋值
a = b = c = 100

# 相当于
a = 100
b = 100
c = 100

# python支持系列解包赋值
a , b = 10 , 20 # 相当于 a = 10, b = 20
```

```python
# 利用解包赋值来交换两个变量的值
a,b = b,a # 将b的值赋给a, 将a的值赋值给b
```

## 比较运算符
| `==` | 等于：判断两个操作数是否相等                     | `a == b` | `False` |
| ---- | ------------------------------------------------ | -------- | ------- |
| `!=` | 不等于：判断两个操作数是否不相等                 | `a != b` | `True`  |
| `>`  | 大于：判断左侧操作数是否大于右侧操作数           | `a > b`  | `True`  |
| `<`  | 小于：判断左侧操作数是否小于右侧操作数           | `a < b`  | `False` |
| `>=` | 大于等于：判断左侧操作数是否大于或等于右侧操作数 | `a >= b` | `True`  |
| `<=` | 小于等于：判断左侧操作数是否小于或等于右侧操作数 | `a <= b` | `False` |

```python
a = 10
b = 20

print(a > b)
print(a < b)
print(a >= b)
print(a <= b)
print(a == b)
print(a != b)
```

## 逻辑运算符
| 运算符 | 描述                                                         | 示例      | 结果（假设 a=True，b=False） |
| ------ | ------------------------------------------------------------ | --------- | ---------------------------- |
| `and`  | 逻辑与：两侧操作数都为`True`时，结果才为`True`，否则为`False` | `a and b` | `False`                      |
| `or`   | 逻辑或：两侧操作数至少有一个为`True`时，结果为`True`，否则为`False` | `a or b`  | `True`                       |
| `not`  | 逻辑非：对操作数取反，`True`变`False`，`False`变`True`       | `not a`   | `False`                      |
